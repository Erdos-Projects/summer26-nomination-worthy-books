```python
from pathlib import Path

def print_tree(directory: Path, prefix: str = "", ignore_list: list = None):
    if ignore_list is None:
        # Define folders or files you want to skip
        ignore_list = [".git", "__pycache__", ".DS_Store"]

    # Filter out any ignored items and sort the remaining ones
    items = [x for x in directory.iterdir() if x.name not in ignore_list]
    items.sort(key=lambda x: (x.is_file(), x.name.lower()))
    
    for i, item in enumerate(items):
        is_last = (i == len(items) - 1)
        connector = "└── " if is_last else "├── "
        
        print(f"{prefix}{connector}{item.name}")
        
        if item.is_dir():
            next_prefix = prefix + ("    " if is_last else "│   ")
            # Pass the ignore list down to recursive calls
            print_tree(item, next_prefix, ignore_list)

# Run the script
target_path = Path(".") 
print(f"{target_path.resolve().name}/")
print_tree(target_path)
```

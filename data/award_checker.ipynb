The purpose of this notebook is to look at a preliminary dataset and compare it with our datasets for Hugo and Locus finalists. We find in this preliminary examination some issues with the dataset such as nonmatching award entries. Certain entry errors are documented below.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
locus = pd.read_csv('locus_nominees.csv')
locus.head()
```

```python
hugo = pd.read_csv('hugo_nominees.csv')
hugo.head()
```

```python
books = pd.read_csv('isfdb-hardcover-combined.csv')
books.title = books.title.apply(lambda x : x.lower())
books.head()
```

```python
hugo.title = hugo.title.apply(lambda x : x.lstrip('“').rstrip('”'))
hugo.title = hugo.title.apply(lambda x : x.lower())
hugo
```

```python
locus.title = locus.title.apply(lambda x : x.strip('“').strip('”'))
locus.title = locus.title.apply(lambda x : x.lower())
locus.head()
```

```python
hugo_mask = books.author.isin(hugo.author) & books.title.isin(hugo.title)
locus_mask = books.author.isin(locus.author) & books.title.isin(locus.title)
```

```python
books['hugo'] = hugo_mask
books['locus'] = locus_mask
```

```python
for title in hugo.title:
    if len(books.loc[books.title == title]) == 0:
        print(title, ':', hugo[hugo.title == title].author.iloc[0])
```

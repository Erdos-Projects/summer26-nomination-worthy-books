```python
tags = auto_tag.load_tags()
```

```python
tags_encoded = transformer.encode(tags)
```

```python
tag_map = {tags[i] : tags_encoded[i] for i in range(len(tags))}
```

```python
for t in tags:
    t_encoded = transformer.encode(t)
    try:
        assert(np.allclose(tag_map[t], t_encoded, atol=1e-6))
    except AssertionError:
        print(tag_map[t] - t_encoded)
        break
```
